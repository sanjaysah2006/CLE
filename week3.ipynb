# Q3_5_invoice

```python
raw_name = " wireless MOUSE             "
raw_sku = "sku-0017"
raw_price = "₹1,299.50"
raw_qty = " 3 "
gst_rate = 18

name = " ".join(raw_name.split()).title()
sku = raw_sku.strip().upper()
price = float(raw_price.replace("₹", "").replace(",", "").strip())
qty = int(raw_qty.strip())

subtotal = round(price * qty, 2)
gst = round(subtotal * gst_rate / 100, 2)
total = round(subtotal + gst, 2)

print("Name:", name)
print("SKU:", sku)
print("Price:", price)
print("Quantity:", qty)
print("Subtotal:", subtotal)
print("GST:", gst)
print("Total:", total)
print(f"{name} | {sku} | {qty} x {price:,.2f} = {subtotal:,.2f} | GST {gst:,.2f} | Total {total:,.2f}") 
```

# Q3_6_shopping_cart

```python
cart = {"P1": 2, "P2": 1, "P3": 5}

cart["P5"] = 1
cart["P2"] += 3
removed_qty = cart.pop("P3")
p9_qty = cart.get("P9", 0)
product_ids = sorted(cart.keys())
total_products = len(cart)
total_units = sum(cart.values())
highest_product = max(cart, key=cart.get)

print("Cart:", cart)
print("Removed quantity:", removed_qty)
print("P9 quantity:", p9_qty)
print("Product IDs:", product_ids)
print("Different products:", total_products)
print("Total units:", total_units)
print("Highest quantity product:", highest_product)
```

# Q3_7_compare_customer_ids

```python
crm_ids = ["C01", "C02", "C03", "C02", "C04", "C07"]
billing_ids = ["C02", "C04", "C05", "C06", "C04"]

unique_crm = list(dict.fromkeys(crm_ids))

crm_set = set(crm_ids)
billing_set = set(billing_ids)

common_ids = sorted(crm_set & billing_set)

crm_only = sorted(crm_set - billing_set)

billing_only = sorted(billing_set - crm_set)

total_unique = len(crm_set | billing_set)

print("A. Unique CRM IDs:", unique_crm)
print("B. Common IDs:", common_ids)
print("C. CRM only:", crm_only)
print("D. Billing only:", billing_only)
print("E. Total unique IDs:", total_unique)
```
